```python
#Import libraries
import os
import httpx
import json
from dotenv import load_dotenv
import gradio as gr
from langchain_core.tools import Tool
from langchain_core.tools import tool
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI
from langchain_openai import AzureChatOpenAI
from typing import Optional, List, Dict, Any
from tool import (
    search_flights_skyline,
    check_exchange_rate,
    get_destination_weather,
    research_vacation_vibe
)
from prompt import build_prompt
from agent import build_agent
import ipywidgets as widgets
from IPython.display import display, Markdown

agent = build_agent()

thread_id = "travel_user_1"

output = widgets.Output(layout={"border": "1px solid gray", "padding": "10px", "height": "350px", "overflow_y": "auto"})
text_box = widgets.Text(placeholder="Type your message here...")
send_button = widgets.Button(description="Send")

def handle_send(_):
    user_input = text_box.value.strip()
    if not user_input:
        return

    text_box.value = ""

    with output:
        display(Markdown(f"**You:** {user_input}"))

    response = agent.invoke(
        {"messages": [("human", user_input)]},
        config={"configurable": {"thread_id": thread_id}}
    )

    assistant_reply = response["messages"][-1].content

    with output:
        display(Markdown(f"**Agent:** {assistant_reply}"))

send_button.on_click(handle_send)

display(output, text_box, send_button)
```
